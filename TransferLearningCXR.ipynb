```python
import os, pathlib
from collections import Counter, defaultdict

P10 = '/content/drive/MyDrive/MIMIC_CXR/mimic-cxr-jpg-2.1.0.physionet.org/files/p10'

print("Tồn tại:", os.path.isdir(P10))

# Đếm folder con cấp 1 (folder bệnh nhân)
patient_dirs = [d for d in os.listdir(P10) if os.path.isdir(os.path.join(P10, d))]
print("Số folder bệnh nhân (cấp 1):", len(patient_dirs))
```

```python
total_files = 0
total_dirs = 0
ext_counter = Counter()
empty_files = []          # file 0 byte
unreadable = []           # file không đọc được

for root, dirs, files in os.walk(P10):
    total_dirs += len(dirs)
    for fn in files:
        total_files += 1
        fp = os.path.join(root, fn)
        ext_counter[pathlib.Path(fn).suffix.lower()] += 1
        try:
            size = os.path.getsize(fp)
            if size == 0:
                empty_files.append(fp)
        except OSError:
            unreadable.append(fp)

print("Tổng số file:", total_files)
print("Tổng số thư mục (mọi cấp):", total_dirs)
print("Phân bố theo đuôi file:", dict(ext_counter))
print("Số file rỗng (0 byte):", len(empty_files))
print("Số file không đọc được:", len(unreadable))
if empty_files[:5]:
    print("Ví dụ file rỗng:", empty_files[:5])
```

```python
import os, pandas as pd

BASE = '/content/drive/MyDrive/MIMIC_CXR/mimic-cxr-jpg-2.1.0.physionet.org'

meta    = pd.read_csv(os.path.join(BASE, 'mimic-cxr-2.0.0-metadata.csv.gz'))
split   = pd.read_csv(os.path.join(BASE, 'mimic-cxr-2.0.0-split.csv.gz'))
testlbl = pd.read_csv(os.path.join(BASE, 'mimic-cxr-2.1.0-test-set-labeled.csv'))

# Prefix bệnh nhân có ảnh trên disk (vd chỉ có 'p10')
have = set(os.listdir(os.path.join(BASE, 'files')))

# Ghép dicom -> subject/study, build đường dẫn ảnh tương đối
meta['pgrp'] = 'p' + meta['subject_id'].astype(str).str[:2]
meta['rel']  = ('files/' + meta['pgrp'] + '/p' + meta['subject_id'].astype(str)
                + '/s' + meta['study_id'].astype(str) + '/' + meta['dicom_id'] + '.jpg')

# Lấy split=test, chỉ dùng ảnh phía trước (frontal AP/PA), thuộc prefix có trên disk
test_df = meta.merge(split[split.split == 'test'][['dicom_id']], on='dicom_id')
test_df = test_df[test_df['ViewPosition'].isin(['AP', 'PA'])]
test_df = test_df[test_df['pgrp'].isin(have)].copy()

# Chỉ giữ ảnh file thật sự tồn tại
test_df['abspath'] = test_df['rel'].apply(lambda r: os.path.join(BASE, r))
test_df = test_df[test_df['abspath'].apply(os.path.isfile)].copy()

# Ghép nhãn test set (theo study_id)
test_df = test_df.merge(testlbl, on='study_id', how='inner')

print('Số ảnh test dùng được:', len(test_df))
print('Số study:', test_df.study_id.nunique())
test_df[['subject_id','study_id','ViewPosition','rel']].head()
```

```python
# Map: tên trong model.pathologies  ->  tên cột trong file test-set-labeled
model_to_label = {
    'Atelectasis'               : 'Atelectasis',
    'Cardiomegaly'              : 'Cardiomegaly',
    'Consolidation'             : 'Consolidation',
    'Edema'                     : 'Edema',
    'Enlarged Cardiomediastinum': 'Enlarged Cardiomediastinum',
    'Fracture'                  : 'Fracture',
    'Lung Lesion'               : 'Lung Lesion',
    'Lung Opacity'              : 'Airspace Opacity',   # <-- sửa ở đây
    'Effusion'                  : 'Pleural Effusion',
    'Pneumonia'                 : 'Pneumonia',
    'Pneumothorax'              : 'Pneumothorax',
}

# Chỉ giữ những cột nhãn thực sự tồn tại trong sample_df (tránh KeyError)
model_to_label = {m: l for m, l in model_to_label.items() if l in sample_df.columns}
print('Số bệnh lý đánh giá được:', len(model_to_label))
print('Các cột nhãn dùng:', list(model_to_label.values()))
```

```python
import numpy as np, torch, skimage.io
import torchxrayvision as xrv
import torchvision
from tqdm import tqdm

transform = torchvision.transforms.Compose([
    xrv.datasets.XRayCenterCrop(),
    xrv.datasets.XRayResizer(224),
])

# Giới hạn để chạy nhanh trên CPU; bỏ dòng .sample nếu muốn chạy toàn bộ
sample_df = test_df.sample(n=min(400, len(test_df)), random_state=0).reset_index(drop=True)

def load_img(path):
    img = skimage.io.imread(path)
    img = xrv.datasets.normalize(img, 255)      # scale về [-1024, 1024]
    if img.ndim == 3: img = img.mean(2)          # -> grayscale
    img = img[None, ...]                          # thêm channel
    return transform(img)

preds, keep = [], []
model.eval()
with torch.no_grad():
    for i, row in tqdm(sample_df.iterrows(), total=len(sample_df)):
        try:
            x = torch.from_numpy(load_img(row['abspath'])).unsqueeze(0).float().to(device)
            out = model(x).cpu().numpy()[0]       # 18 xác suất
            preds.append(out); keep.append(i)
        except Exception as e:
            print('skip', row['abspath'], e)

preds = np.array(preds)
sample_df = sample_df.loc[keep].reset_index(drop=True)
print('Đã dự đoán:', preds.shape)
```

```python
from sklearn.metrics import roc_auc_score
import numpy as np, pandas as pd

pathologies = model.pathologies
results = {}
for m_name, l_name in model_to_label.items():
    if l_name not in sample_df.columns:      # phòng vệ
        continue
    idx = pathologies.index(m_name)
    y = sample_df[l_name].values             # 1=có, 0=không, -1/NaN=không chắc
    mask = np.isin(y, [0.0, 1.0])            # chỉ dùng nhãn chắc chắn
    if mask.sum() > 0 and len(np.unique(y[mask])) == 2:
        results[m_name] = roc_auc_score(y[mask], preds[mask, idx])

res = pd.Series(results).sort_values(ascending=False)
print(res.to_string())
print('\nMacro AUC:', round(res.mean(), 4), '| số bệnh lý:', len(res))
```
